```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.datasets import load_svmlight_file
import numpy as np
```

```python
def accuracy_tpr_tnr(y_real_predict):
    accuracy = 0
    tp = 0
    p = 0
    tn = 0
    n = 0
    for i in range(len(y_real_predict)):  # Iterate over each row
        if y_real_predict[i, 0] == y_real_predict[i, 1]:  # Compare the two columns
            accuracy+=1
            if y_real_predict[i, 0] == 1:
                tp+=1
            else:
                tn+=1
        if y_real_predict[i, 0] == 1:
            p+=1
        else:
            n+=1

    return accuracy/len(y_real_predict), tp/p if  p > 0 else 0, tn/n if n > 0 else 0
```

```python
trainfile='../../SVM_model/cds/training_features_noEarliest.tsv'
X_sparse, y_train = load_svmlight_file(trainfile)
X_train = X_sparse.toarray()
```

```python
clf = RandomForestClassifier(class_weight={1: 1, -1: 16}, max_depth=4)
#clf = RandomForestClassifier(class_weight='balanced_subsample', max_depth=2)
clf.fit(X_train, y_train)
y_train_predict = clf.predict(X_train)
y_train_real_predict = np.column_stack((y_train, y_train_predict))
np.savetxt("y_train_real_predict.tsv", y_train_real_predict, fmt='%d', delimiter='\t')
print(accuracy_tpr_tnr(y_train_real_predict))
```

```python
testfile='../../SVM_model/cds/testing_features_Earliest.tsv'
X_sparse, y_test = load_svmlight_file(testfile)
X_test = X_sparse.toarray()
y_test_predict = clf.predict(X_test)
y_test_real_predict = np.column_stack((y_test, y_test_predict))
np.savetxt("y_test_real_predict.tsv", y_test_real_predict, fmt='%d', delimiter='\t')
print(accuracy_tpr_tnr(y_test_real_predict))
```
